# Prototype → `santa`

The original exploration notebook lived under `prototype/`. The pipeline is now `santa.card.make_card`.
Playgrounds for each media role: `01_text_to_image.ipynb` … `04_video_generation.ipynb`.
Eval notebooks: `11_gift_rec_eval.ipynb` … `14_benchmark_llm.ipynb`.

Set `SANTA_EVAL_FAKE=1` to run this cell without endpoints.


```python
import os
from pathlib import Path

from santa.card import make_card
from santa.config import Settings
from santa.eval import FakeEvalImage, FakeEvalLLM
from santa.models import adapter_for
from santa.prompts import gift_recommendation_prompt
from santa.schemas import KidProfile

kid = KidProfile(name="Emma", age=7, wishlist=["a telescope"])
print(gift_recommendation_prompt(kid)[:120], "…")

out = Path("out") / "prototype-demo"
if os.environ.get("SANTA_EVAL_FAKE", "1").strip() in {"1", "true", "yes"}:
    llm, image = FakeEvalLLM(), FakeEvalImage()
else:
    settings = Settings.load()
    llm, image = adapter_for("llm", settings), adapter_for("image", settings)

run = make_card(kid, Settings.load(), out_root=out, llm=llm, image=image)
print(run.card.png_path)
print(run.card.html_path)
```
